## 1. Introduction
Linear Regression is a predictive modeling technique that investigates the relationship between a dependent (target) variable and one or more independent (predictor) variables.
It quantifies this relationship by fitting a straight line (the "best fit line") through the data points.

**Mathematical Representation:**
\(h_\theta(x) = \theta_0 + \theta_1 x\)

*   **\(h_\theta(x)\)**: The predicted output (e.g., Distance)
*   **\(x\)**: The input variable (e.g., Speed)
*   **\(\theta_1\) (Slope)**: Determines how much \(x\) impacts the output.
*   **\(\theta_0\) (Intercept)**: The constant value when \(x = 0\).

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## 2. The Dataset
We will use the exact dataset from Page 13 of the lecture slides. We are trying to find the best fit line using the **Ordinary Least Squares (OLS)** method.

| x (Input) | hθ(x) (Output) |
| :--- | :--- |
| 1 | 3 |
| 2 | 4 |
| 3 | 2 |
| 4 | 4 |
| 5 | 5 |

```python
# 1. Create the dataset
data = {'x': [1, 2, 3, 4, 5],
        'y': [3, 4, 2, 4, 5]}
df = pd.DataFrame(data)

# 2. Calculate Means
x_mean = df['x'].mean()
y_mean = df['y'].mean()

print(f"Mean of x: {x_mean}")
print(f"Mean of y: {y_mean}")

# 3. Calculate Theta 1 (Slope) using Least Squares Formula
numerator = ((df['x'] - x_mean) * (df['y'] - y_mean)).sum()
denominator = ((df['x'] - x_mean)**2).sum()
theta_1 = numerator / denominator

# 4. Calculate Theta 0 (Intercept)
theta_0 = y_mean - (theta_1 * x_mean)

print("\n--- Results ---")
print(f"Theta 1 (Slope): {theta_1}")
print(f"Theta 0 (Intercept): {theta_0}")
print(f"Regression Equation: h(x) = {theta_0:.1f} + {theta_1:.1f}x")
```

## 3. Visualizing the Best Fit Line
Now we will plot the actual data points and draw the regression line we just calculated. We will also visualize the **residuals** (the errors) which are the vertical lines between the actual points and the line.

```python
# 5. Create the plot
plt.figure(figsize=(10, 6))

# Plot actual data points
plt.scatter(df['x'], df['y'], color='red', s=100, label='Actual Data', zorder=5)

# Calculate the regression line
x_values = np.array([1, 2, 3, 4, 5, 6])
y_predicted = theta_0 + (theta_1 * x_values)

# Plot the regression line
plt.plot(x_values, y_predicted, color='blue', linestyle='--', linewidth=2, label=f'Regression Line: h(x) = {theta_0} + {theta_1}x', zorder=4)

# Plot residuals (vertical lines)
for i in range(len(df)):
    actual_y = df['y'][i]
    pred_y = theta_0 + (theta_1 * df['x'][i])
    plt.plot([df['x'][i], df['x'][i]], [actual_y, pred_y], color='gray', linestyle=':', zorder=3)

plt.title('Linear Regression: Best Fit Line & Residuals', fontsize=14)
plt.xlabel('x (Input)', fontsize=12)
plt.ylabel('hθ(x) (Output)', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Conclusion
Using the Ordinary Least Squares method, we successfully calculated the regression line for the dataset:
**\(h_\theta(x) = 2.4 + 0.4x\)**

Linear regression provides a highly transparent, comprehensible, and computationally efficient way to model continuous data, making it a foundational tool in machine learning.
